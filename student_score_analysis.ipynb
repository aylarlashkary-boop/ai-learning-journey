# Student Score Analysis

A small Python exercise for analyzing a list of student scores.

## Learning objectives

- Calculate the average score.
- Count scores that meet the passing threshold.
- Find the highest score.
- Handle an empty list explicitly.
- Practice writing a reusable function.

## Problem

Given a list of scores, calculate the average, count how many scores are at least `10`, and find the highest score. If the list is empty, report that input as invalid.

The passing threshold in this exercise is **10**.

## 1. Implement the analysis function

The function below **returns** its results instead of printing them. Returning values makes the function easier to reuse in other parts of a program.

An empty list has no average or highest score, so the function raises a `ValueError` with a clear message. This is an intentional way to tell the caller that the input is not valid for this task.

````python
def analyze_scores(scores):
    if not scores:
        raise ValueError("Please provide at least one score.")

    average_score = sum(scores) / len(scores)
    passing_students = sum(score >= 10 for score in scores)
    highest_score = max(scores)

    return {
        "average_score": average_score,
        "passing_students": passing_students,
        "highest_score": highest_score,
    }
````

## 2. Run the function

Try the function with the example data. The returned dictionary stores each result under a descriptive key.

````python
scores = [12, 18, 7, 15, 9, 20, 14]
results = analyze_scores(scores)

print(f"Average score: {results['average_score']:.2f}")
print(f"Passing students: {results['passing_students']}")
print(f"Highest score: {results['highest_score']}")
````

### Expected output

```text
Average score: 13.57
Passing students: 5
Highest score: 20
```

The list contains five scores that are at least `10`: `12`, `18`, `15`, `20`, and `14`.

## 3. Check the empty-list case

An empty list should not be analyzed. We can catch the specific `ValueError` and display its message.

````python
try:
    analyze_scores([])
except ValueError as error:
    print(f"Input error: {error}")
````

## What I learned

- How to organize a task inside a reusable function.
- How `sum()`, `len()`, and `max()` help analyze a list.
- How to count matching items with a Boolean condition.
- Why returning results can be more reusable than printing inside a function.
- How to handle an empty list with a specific exception.

## Limitations

This is a small practice exercise. It assumes that `scores` contains numeric values and does not validate score ranges or other input types.
